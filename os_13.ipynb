### Python equivalent of the C code

```python
import os

def run_c_code_equivalent():
    # Simulate user input
    regno = int(input("Enter Register Number (C equivalent): "))
    name = input("Enter Name (C equivalent): ")

    # Write to student.dat
    with open("student.dat", "w") as fp:
        fp.write(f"{regno} {name}\n")

    # Read from student.dat
    s_regno = None
    s_name = None
    if os.path.exists("student.dat"):
        with open("student.dat", "r") as fp:
            line = fp.readline().strip()
            parts = line.split(' ', 1)
            if len(parts) == 2:
                s_regno = int(parts[0])
                s_name = parts[1]

    # Print record details
    print("\nRecord Details (C equivalent)")
    if s_regno is not None and s_name is not None:
        print(f"Register Number : {s_regno}")
        print(f"Name : {s_name}")
    else:
        print("Could not read student record.")

run_c_code_equivalent()
```

### Python equivalent of the Shell script

```python
import os

def run_shell_script_equivalent():
    # Simulate user input
    regno_shell = input("Enter Register Number (Shell equivalent): ")
    name_shell = input("Enter Name (Shell equivalent): ")

    # Write to student.txt
    with open("student.txt", "w") as f:
        f.write(f"{regno_shell} {name_shell}\n")

    # Print contents of student.txt
    print("Contents of File (Shell equivalent)")
    if os.path.exists("student.txt"):
        with open("student.txt", "r") as f:
            print(f.read())
    else:
        print("student.txt not found.")

run_shell_script_equivalent()
```
